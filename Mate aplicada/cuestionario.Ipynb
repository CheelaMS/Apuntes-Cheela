```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve
import math
```

```python
def s(x):
    return -1.6*x**2 + 1.3*x + 141.3

def r(x):
    return 1.4*x**2 - 5.6*x + 12.6

x = np.arange(-100, 200, 0.1)

plt.plot(x, s(x), color = 'r', label = 'g(x)')
plt.plot(x, r(x), color = 'g', label = 'f(x)')
plt.ylim(0, 200)
plt.xlim(-10, 10)

plt.title('Comparación entre las funciones g(x) y f(x)')

plt.grid(True)
plt.legend()
plt.show()

def inter(x):
  return s(x)-r(x)

x = np.linspace(-10, 20, 2)

#np.around() permite redondear los valores a 2 decimales y elimina los restantes
solucion = np.around(fsolve(inter, x), decimals = 2)

#np.unique() permite eliminar las soluciones repetidas
interseccion = np.unique(solucion)

print(f'Los puntos de intersección de las funciones están en x = {interseccion}')
```

```python
def I(x):
    return 50*x + 0.5*x**2

def C2(x):
    return 10*x + 450

def inter2(x):
  return I(x)-C2(x)

x = np.linspace(0, 1000, 2)

#np.around() permite redondear los valores a 2 decimales y elimina los restantes
solucion = np.around(fsolve(inter2, x), decimals = 2)

#np.unique() permite eliminar las soluciones repetidas
interseccion = np.unique(solucion)

print(f'Los puntos de intersección de las funciones están en x = {interseccion}')


```
